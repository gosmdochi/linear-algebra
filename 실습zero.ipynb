### 반복문, 조건문

```python
contents = ["Regression", "Classification", "SYM", "Clustering", "Dimension reduction",
            "NN", "CNN", "AE", "GAN", "RNN"]

for con in contents:
  if con in ["Regression", "Classification", "SYM", "Clustering", "Dimension reduction"]:
    print("%s은(는) 기계학습 내용입니다." %con)
  elif con in ["CNN"]:
    print("%s은(는) convolutional neural network 내용입니다." %con)
  else:
    print("%s은(는) 심층학습 내용입니다." %con)
```

### 반복문과 인덱스

```python
for (i, con) in enumerate(contents):
  print("[%d%d]: %s" %(i, len(contents), con))
```

### 클래스

```python
class Student:
  # 생성자
  def __init__(self, name):
    self.name = name

  # 메써드
  def study(self, hard = False):
    if hard:
      print("%s 학생은 열심히 공부합니다." %self.name)

    else:
      print("%s 학생은 공부합니다." %self.name)

s = Student('Heekyung')
s.study()
s.study(hard = True)
```

### 라이브러리(패키지) 로드

```python
import numpy as np
```

### 프린트

```python
def print_val(x):
  print("Type:", type(x))
  print("Shape:", x.shape)
  print("값:\n", x)
  print(" ")
```

rank 1 np array

```python
x = np.array([1, 2, 3])
print_val(x)

x[0] = 5
print_val(x)
```

### rank 2 np array

```python
y = np.array([[1, 2, 3], [4, 5, 6]])
print_val(y)
```

### rank 2 ones

```python
a = np.ones((3, 2))
print_val(a)
```

### rank 2 zeros

```python
a = np.zeros((2, 2))
print_val(a)
```

### rank 2 단위행렬(identity matrix)

```python
a = np.eye(3, 3)
print_val(a)
```

### 랜덤 행렬(uniform: 0~1 사이 모든 값들이 나올 확률이 같음

```python
a = np.random.random((4, 4))
print_val(a)
```

### 랜덤 행렬(Gaussian: 0을 평균으로 하는 가우시안 분포를 따르는 랜덤값

```python
a = np.random.randn(4, 4)
print_val(a)
```
